```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_absolute_error

df=pd.read_csv('/kaggle/input/datasets/dudefacts/kerala-real-estates/data.csv')

print(df.shape)
print(df.describe())
print(df.isnull().sum())
```

```python
features = ['age', 'amneties', 'area', 'bathrooms', 'balconies', 
            'atmDistance', 'hospitalDistance', 'schoolDistance', 
            'shoppingDistance', 'restrauntDistance', 'status']
X=df[features]
y=df['price']

train_X,val_X,train_y,val_y= train_test_split(X,y,random_state=1)

model = DecisionTreeRegressor(random_state=1)
model.fit(train_X,train_y)

preds = model.predict(val_X)
mae = mean_absolute_error(preds,val_y)
print("MAE:",mae)
```

```python
from sklearn.ensemble import RandomForestRegressor

model = RandomForestRegressor(n_estimators=200,random_state=1)
model.fit(train_X, train_y)

preds = model.predict(val_X)
mae = mean_absolute_error(val_y, preds)
print("MAE:", mae)
```

```python
# Remove outliers
df_clean = df[df['area'] < 10000]
print(df_clean.shape)
```

```python
X = df_clean[features]
y = df_clean['price']

train_X, val_X, train_y, val_y = train_test_split(X, y, random_state=1)

model = RandomForestRegressor(n_estimators=200, random_state=1)
model.fit(train_X, train_y)

preds = model.predict(val_X)
mae = mean_absolute_error(val_y, preds)
print("MAE:", mae)
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error

for trees in [100, 200, 300, 500]:
    model = RandomForestRegressor(n_estimators=trees, random_state=1)
    model.fit(train_X, train_y)
    preds = model.predict(val_X)
    mae = mean_absolute_error(val_y, preds)
    print(f"Trees: {trees} → MAE: {mae:,.0f}")
```

```python
for depth in [5, 10, 20, 30, None]:
    model = RandomForestRegressor(n_estimators=100, max_depth=depth, random_state=1)
    model.fit(train_X, train_y)
    preds = model.predict(val_X)
    mae = mean_absolute_error(val_y, preds)
    print(f"Depth: {depth} → MAE: {mae:,.0f}")
```
